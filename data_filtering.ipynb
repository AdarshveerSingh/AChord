## Data Filtering and Export

This section filters the dataset based on selected criteria and saves the result for further processing.

---

### Data Loading

- Loads the dataset from a CSV file into a structured format  
- Displays column names to verify available features  

---

### Filtering Criteria

- Retains only entries that meet the following conditions:
  - Year is greater than or equal to 1999  
  - Popularity is greater than or equal to 50  

---

### Result Inspection

- Displays the shape of the filtered dataset  
- Provides a preview of the filtered data  

---

### Data Export

- Saves the filtered dataset to a new CSV file  
- Ensures the index is not included in the output file  

---

### Purpose

- Reduces dataset size for efficient processing  
- Focuses on relatively recent and popular tracks  
- Prepares clean input for downstream tasks such as tag extraction and recommendation  

```python
import pandas as pd

df = pd.read_csv("songs.csv")


print(df.columns)

filtered_df = df[
    (df['year'] >= 1999) &
    (df['popularity'] >= 50)
]

print("Filtered shape:", filtered_df.shape)
filtered_df.head()

filtered_df.to_csv("filtered_songs.csv", index=False)

print("✅ Saved as filtered_songs.csv")
```

## Extracting Main Artist

This section processes the artist information to extract a single primary artist for each track.

---

### Data Loading

- Loads the filtered dataset from a CSV file  
- Prepares it for further transformation  

---

### Artist Extraction Logic

- The artist field may contain multiple artists stored as a list-like string  
- Parses the string representation into a structured list  
- Selects the first artist as the main artist  
- If parsing fails, retains the original value  

---

### Column Transformation

- Creates a new column representing the main artist  
- Ensures each track is associated with a single artist  

---

### Result Preparation

- Selects only the required columns:
  - Track name  
  - Main artist  

---

### Data Export

- Saves the processed data into a new CSV file  
- Removes the index for cleaner output  

---

### Purpose

- Simplifies multi-artist data into a single representative artist  
- Prepares consistent input for downstream tasks such as tag fetching and recommendation  

```python
import pandas as pd
import ast

df = pd.read_csv("filtered_songs.csv")

def get_main_artist(x):
    if isinstance(x, str):
        try:
            artists = ast.literal_eval(x)
            if isinstance(artists, list) and len(artists) > 0:
                return artists[0]
        except:
            return x
    return x

df['main_artist'] = df['artists'].apply(get_main_artist)

result = df[['name', 'main_artist']]

result.to_csv("songs_main_artist.csv", index=False)

result.head()
```
