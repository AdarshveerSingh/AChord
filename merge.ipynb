## Merging Songs with Cleaned Genres

This section combines the original song dataset with the processed genre information.

---

### Data Loading

- Loads the original songs dataset  
- Loads the dataset containing filtered and cleaned genre tags  

---

### Text Normalization

- Standardizes track names and artist names in both datasets  
- Converts text to lowercase  
- Removes punctuation and extra spaces  
- Ensures consistent formatting for accurate matching  

---

### Merge Strategy

- Merges the two datasets using:
  - Track name  
  - Artist name  
- Uses normalized versions of these fields to improve match accuracy  
- Performs a left join to retain all original song entries  

---

### Data Cleanup

- Removes temporary normalized columns after merging  
- Keeps the dataset clean and readable  

---

### Final Output

- Produces a unified dataset containing songs with their corresponding cleaned genres  
- Saves the result to a new file  

---

### Purpose

- Aligns genre data with the original dataset  
- Resolves inconsistencies caused by formatting differences  
- Prepares a complete dataset for modeling and recommendation tasks  

```python
import pandas as pd
import re

# ---------- LOAD ----------
songs = pd.read_csv("songs.csv")
genres = pd.read_csv("songs_final_filtered.csv")

# ---------- NORMALIZE ----------
def normalize(x):
    x = str(x).lower()
    x = re.sub(r'[^\w\s]', '', x)
    x = re.sub(r'\s+', ' ', x).strip()
    return x

songs['name_clean'] = songs['name'].apply(normalize)
songs['artist_clean'] = songs['main_artist'].apply(normalize)

genres['name_clean'] = genres['name'].apply(normalize)
genres['artist_clean'] = genres['main_artist'].apply(normalize)

# ---------- MERGE ----------
merged = songs.merge(
    genres[['name_clean', 'artist_clean', 'genres_filtered']],
    on=['name_clean', 'artist_clean'],
    how='left'
)

# ---------- CLEAN UP ----------
merged.drop(columns=['name_clean', 'artist_clean'], inplace=True)

# ---------- SAVE ----------
merged.to_csv("songs_with_genres_final.csv", index=False)

print("Done. Genres merged using name + artist.")
```
